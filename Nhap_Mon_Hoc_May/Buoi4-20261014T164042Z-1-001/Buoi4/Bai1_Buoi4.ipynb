```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB

# 1. Tạo tập dữ liệu từ bảng trong ảnh
data = {
    'Age': ['youth', 'youth', 'middle', 'senior', 'senior', 'senior', 'middle', 'youth', 'youth', 'senior', 'middle', 'youth', 'middle', 'senior', 'youth'],
    'Income': ['high', 'high', 'high', 'medium', 'low', 'low', 'low', 'medium', 'low', 'medium', 'medium', 'medium', 'high', 'medium', 'low'],
    'Student': ['no', 'no', 'no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'yes', 'no', 'yes', 'yes', 'no', 'no'],
    'Credit': ['fair', 'excellent', 'fair', 'fair', 'fair', 'excellent', 'excellent', 'fair', 'fair', 'fair', 'excellent', 'excellent', 'fair', 'excellent', 'fair'],
    'BuyLaptop': ['no', 'no', 'yes', 'yes', 'yes', 'no', 'yes', 'yes', 'yes', 'yes', 'yes', 'yes', 'yes', 'no', 'no']
}

df = pd.DataFrame(data)

# 2. Mã hóa dữ liệu dạng chuỗi sang dạng số để đưa vào mô hình
features = ['Age', 'Income', 'Student', 'Credit']
target = 'BuyLaptop'

encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(df[features])
y = df[target]

# 3. Huấn luyện mô hình Naive Bayes (thuật toán phổ biến nhất cho bài toán này)
model = CategoricalNB()
model.fit(X_encoded, y)

# 4. Dự đoán cho mẫu X = (youth, medium, yes, fair)
sample_X = [['youth', 'medium', 'yes', 'fair']]
sample_encoded = encoder.transform(sample_X)

prediction = model.predict(sample_encoded)
probabilities = model.predict_proba(sample_encoded)

# 5. In kết quả
print("=== KẾT QUẢ DỰ ĐOÁN ===")
print(f"Mẫu cần dự đoán: X = {sample_X[0]}")
print(f"Kết quả dự đoán BuyLaptop: {prediction[0]}")
print(f"Xác suất (no, yes): {probabilities[0]}")ư
```
