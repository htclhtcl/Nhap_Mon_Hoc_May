```python
# 1. Nhập thư viện cần thiết
import pandas as pd
import numpy as np

# 2. Đọc dữ liệu từ file
df = pd.read_csv('sample_data.csv')
print("--- Dữ liệu ban đầu (head) ---")
print(df.head())

# 3. Kiểm tra thông tin dữ liệu
print("\n--- Thông tin dữ liệu (info) ---")
df.info()

print("\n--- Thống kê mô tả (describe) ---")
print(df.describe(include='all'))

# 4. Kiểm tra dữ liệu thiếu
print("\n--- Số lượng giá trị thiếu trên mỗi cột ---")
print(df.isnull().sum())

# Tính tỉ lệ phần trăm thiếu
missing_ratio = df.isnull().mean() * 100
print("\n--- Tỉ lệ % giá trị thiếu ---")
print(missing_ratio)

# 5. Xử lý dữ liệu thiếu
# 5.1. Xóa các hàng có giá trị thiếu (Chú ý: Nếu dùng cách này sẽ xóa cả dataframe, ở dưới dùng 5.2 theo bài thực hành)
# df.dropna(inplace=True)

# 5.2. Điền giá trị mặc định / trung bình / yếu vị
# Điền giá trị trung bình cho cột 'Age'
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Điền giá trị thường gặp nhất (mode) cho cột 'Gender' (nếu có thiếu)
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)

# 6. Xử lý dữ liệu trùng lặp
print("\n--- Số lượng bản ghi trùng ---")
print(df.duplicated().sum())

# Xóa bản ghi trùng
df.drop_duplicates(inplace=True)

# 7. Chuẩn hóa dữ liệu
# Chuẩn hóa chữ: ví dụ chuyển về chữ thường
df['Gender'] = df['Gender'].str.lower()

# 8. Chuyển đổi kiểu dữ liệu nếu cần
# Chuyển cột 'JoinDate' (trong bảng ảnh 1) sang kiểu datetime
df['JoinDate'] = pd.to_datetime(df['JoinDate'], errors='coerce')

# 9. Lưu lại dữ liệu đã làm sạch
df.to_csv('sample_data_cleaned.csv', index=False)

print("\n--- Dữ liệu sau khi làm sạch ---")
print(df)
print("\nĐã lưu file làm sạch thành công: 'sample_data_cleaned.csv'")
```
