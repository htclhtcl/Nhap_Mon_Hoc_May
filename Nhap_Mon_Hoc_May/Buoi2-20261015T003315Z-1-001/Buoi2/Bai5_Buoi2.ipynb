```python
# ==========================================
# BÀI 5: CHUẨN HÓA CHUỖI VÀ BIẾN PHÂN LOẠI
# ==========================================
import pandas as pd
import numpy as np
import re
from google.colab import drive

# 1. Mount Drive và Khởi tạo df_clean
drive.mount('/content/drive')
file_path = '/content/drive/MyDrive/Nhap_Mon_Hoc_May/Buoi2/Du_lieu_thuc_hanh_chuan_bi_du_lieu.xlsx'

df_raw = pd.read_excel(file_path, sheet_name='Du_lieu_tho')

# Tạo DataFrame làm sạch chính thức theo yêu cầu bài học
df_clean = df_raw.copy()

def print_row_count(step_name):
    print(f"-> Số dòng sau bước [{step_name}]: {len(df_clean)}")

print("=== BẮT ĐẦU THỰC HIỆN BÀI 5 ===")
print(f"Số dòng ban đầu: {len(df_clean)}\n")


# ------------------------------------------
# Yêu cầu 20: Loại khoảng trắng thừa ở đầu, cuối và giữa chuỗi
# ------------------------------------------
str_cols = df_clean.select_dtypes(include=['object']).columns

for col in str_cols:
    df_clean[col] = df_clean[col].astype(str).apply(
        lambda x: re.sub(r'\s+', ' ', x.strip()) if x != 'nan' else np.nan
    )

print_row_count("20. Loại khoảng trắng thừa")


# ------------------------------------------
# Yêu cầu 21: Chuẩn hóa Category về đúng 5 nhóm trong từ điển dữ liệu
# ------------------------------------------
# Chuyển về chữ thường để đồng nhất trước khi ánh ánh xạ
df_clean['Category'] = df_clean['Category'].str.lower()

# Định nghĩa bảng ánh xạ các biến thể về 5 nhóm chuẩn
category_mapping = {
    'electronics': 'Electronics',
    'elec': 'Electronics',
    'dientu': 'Electronics',
    'fashion': 'Fashion',
    'thoitrang': 'Fashion',
    'home & kitchen': 'Home & Kitchen',
    'home': 'Home & Kitchen',
    'kitchen': 'Home & Kitchen',
    'beauty': 'Beauty',
    'lamdep': 'Beauty',
    'sports': 'Sports',
    'thethao': 'Sports'
}

df_clean['Category'] = df_clean['Category'].map(category_mapping).fillna(df_clean['Category'].str.title())
print_row_count("21. Chuẩn hóa Category")


# ------------------------------------------
# Yêu cầu 22: Chuẩn hóa Seller_City về tên thành phố chuẩn
# ------------------------------------------
df_clean['Seller_City_temp'] = df_clean['Seller_City'].str.lower()

city_mapping = {
    'hn': 'Hà Nội',
    'ha noi': 'Hà Nội',
    'hà nội': 'Hà Nội',
    'hcm': 'TP. Hồ Chí Minh',
    'tp.hcm': 'TP. Hồ Chí Minh',
    'ho chi minh': 'TP. Hồ Chí Minh',
    'tp hồ chí minh': 'TP. Hồ Chí Minh',
    'dn': 'Đà Nẵng',
    'da nang': 'Đà Nẵng',
    'đà nẵng': 'Đà Nẵng'
}

df_clean['Seller_City'] = df_clean['Seller_City_temp'].map(city_mapping).fillna(df_clean['Seller_City'])
df_clean.drop(columns=['Seller_City_temp'], inplace=True)

print_row_count("22. Chuẩn hóa Seller_City")


# ------------------------------------------
# Yêu cầu 23: Chuẩn hóa Has_Image về 1 và 0 (Kèm giải thích)
# ------------------------------------------
# Ánh xạ các giá trị đúng/có/1 -> 1, các giá trị sai/không/0 -> 0
has_image_map = {
    '1': 1, 1: 1, 'true': 1, True: 1, 'yes': 1, 'có': 1,
    '0': 0, 0: 0, 'false': 0, False: 0, 'no': 0, 'không': 0
}

df_clean['Has_Image'] = df_clean['Has_Image'].str.lower().map(has_image_map)

# Xử lý giá trị lạ hoặc thiếu: Điền giá trị phổ biến nhất (mode) = 0 hoặc 1
mode_val = df_clean['Has_Image'].mode()[0]
df_clean['Has_Image'] = df_clean['Has_Image'].fillna(mode_val).astype(int)

print_row_count("23. Chuẩn hóa Has_Image")

print("\n[GIẢI THÍCH YÊU CẦU 23]:")
print("- Các giá trị đại diện cho trạng thái 'có' (1, true, yes, có) được ánh xạ về 1.")
print("- Các giá trị đại diện cho trạng thái 'không' (0, false, no, không) được ánh xạ về 0.")
print(f"- Các giá trị thiếu (NaN) hoặc lạ không khớp quy tắc được quy đổi về giá trị xuất hiện nhiều nhất (mode = {mode_val}) để đảm bảo tính toàn vẹn dữ liệu nhị phân.")


# ------------------------------------------
# Yêu cầu 24: Kiểm tra lại value_counts() để chứng minh số nhóm đã giảm
# ------------------------------------------
print("\n--- 24. GIÁ TRỊ TẦN SUẤT SAU KHI CHUẨN HÓA (ĐÃ GIẢM SỐ NHÓM) ---")

print("\n* Cột Category sau chuẩn hóa:")
print(df_clean['Category'].value_counts(dropna=False))

print("\n* Cột Seller_City sau chuẩn hóa:")
print(df_clean['Seller_City'].value_counts(dropna=False))

print("\n* Cột Has_Image sau chuẩn hóa:")
print(df_clean['Has_Image'].value_counts(dropna=False))
```
